Lab 5.1: Simple Contact Book Manager

```python
contacts = {}

while True:
    print("\n===== Contact Book Menu =====")
    print("1. Add/Update Contact")
    print("2. View Contact Details")
    print("3. List All Contacts")
    print("4. Delete Contact")
    print("5. Exit")

    choice = input("Enter your choice (1-5): ")

    if choice == "1":
        name = input("Enter contact name: ").strip().lower()

        phone = input("Enter phone number: ")
        email = input("Enter email: ")

        contacts[name] = {
            "phone": phone,
            "email": email
        }

        print(f"Contact '{name.title()}' added/updated successfully!")

    elif choice == "2":
        name = input("Enter contact name: ").strip().lower()

        contact = contacts.get(name)

        if contact:
            print("\nContact Details")
            print("Name :", name.title())
            print("Phone:", contact["phone"])
            print("Email:", contact["email"])
        else:
            print("Contact not found.")

    elif choice == "3":
        if len(contacts) == 0:
            print("No contacts available.")
        else:
            print("\n===== All Contacts =====")
            for name, details in contacts.items():
                print(f"Name : {name.title()}")
                print(f"Phone: {details['phone']}")
                print(f"Email: {details['email']}")
                print("--------------------")

    elif choice == "4":
        name = input("Enter contact name to delete: ").strip().lower()

        removed = contacts.pop(name, None)

        if removed:
            print(f"Contact '{name.title()}' deleted.")
        else:
            print("Contact not found.")

    # 5. Exit
    elif choice == "5":
        print("Exiting program...")
        break

    else:
        print("Invalid choice. Please enter 1-5.")
```
